```python
cgpa = float(input("Enter your cgpa :"))
grade="O"
if cgpa>=10:
    grade="O"
elif cgpa>9:
    grade="E"
elif cgpa>8:
    grade="A"
elif cgpa>7:
    grade="B"
elif cgpa>6:
    grade="C"
else :
    grade="D"
    
if grade !="D":
    print("your grade is",grade)
else:
    print("You failed")
```
