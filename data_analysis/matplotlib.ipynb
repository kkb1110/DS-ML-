```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]
plt.plot(x,y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic XY plot')
plt.show()
```

Customizing line plot

```python


x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y, color='orange', marker='o', linestyle='--', linewidth=1.5, markersize=4)
#possible marker styles: 'o', 's', '^', 'D', '*', '+', 'x'
#possible linestyles: '-', '--', '-.', ':'
#linewidth controls the thickness of the line
#markersize controls the size of the markers    
#possible colors: 'blue', 'green', 'red', 'cyan', 'magenta', 'yellow', 'black', 'orange', etc.
plt.grid(True)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Customized plot')
plt.show()
```

Multiple plots

```python
### multiple plots
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [2, 3, 5, 7, 11]

### Defining the overall figure
plt.figure(figsize=(10, 6))
plt.suptitle('Multiple Subplots Example', fontsize=16)

### 1st subplot
plt.subplot(2,2,1)
plt.plot(x, y1, label='y = x^2', color='blue', marker='o')
plt.title('Plot 1')

### 2nd subplot
plt.subplot(2,2,2)
plt.plot(x, y2, label='y = prime numbers', color='red', marker='s')
plt.title('Plot 2')

### 3rd subplot
plt.subplot(2,2,3)
plt.plot(y1,y2, label='y = x^2', color='orange', marker='o')
plt.xlabel('X------axis')
plt.ylabel('Y-----axis')
plt.title('Plot 3')

### 4th subplot
plt.subplot(2,2,4)
plt.plot(y1, y2,  color='green', marker='^')
plt.xlabel('X---axis')
plt.ylabel('Y---axis')
plt.title('Plot 4')


# plt.tight_layout(rect=[0, 0.03, 1, 0.95])
# plt.xlabel('X-axis')
# plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

Bar plots

```python
categories = ['A', 'B', 'C', 'D']
values = [10, 15, 7, 12]

### Creating the bar plot
# plt.bar(categories, values, color=['blue', 'orange', 'green', 'red'])
plt.bar(categories, values, color='orange', edgecolor='black', alpha=0.9, width=0.8, align='center', hatch='-.')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Plot Example')
plt.show()
```

Histograms

```python
### sample data
data = [12, 15, 13, 17, 19, 21, 22, 25, 30, 28, 27, 35, 40, 38,15,45]
### Creating the histogram
# a histogram is used to represent the distribution of numerical data by dividing the data into bins or intervals

plt.hist(data, bins=5, color='orange', edgecolor='black', alpha=0.7)
plt.xlabel('Value Ranges')
plt.ylabel('Frequency') 
plt.title('Histogram Example')
```

Scatter plot

```python
### create scatter plot sample data
x = [5, 7, 8, 5, 6, 7, 8, 9, 6, 5, 7, 8]
y = [10, 12, 15, 10, 11, 13, 14, 16, 11, 10, 12, 15]

plt.scatter(x, y, color='green', marker='o', alpha=0.6, edgecolors='black', s=110)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot Example')
plt.grid(True)
plt.show()
```

Pie chart

```python
### Pie chart
labels=['A','B','C','D']
sizes=[25,35,20,20]
colors=['gold','lightcoral','lightskyblue','lightgreen']
explode=(0,0.1,0,0)  # explode 1st slice

### create pie chart
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True,explode=explode)
#autopct displays the percentage value on each slice
```

```python
import pandas as pd
sales_data_df=pd.read_csv("100 Sales Records.csv")
sales_data_df.head()


```

```python
## plot total sales(unit sold) by item type 
import matplotlib.pyplot as plt

# Group the data by 'Item Type' and sum the 'Quantity Sold'
item_sales = sales_data_df.groupby('Item Type')['Units Sold'].sum()

plt.bar(item_sales.index, item_sales.values, color='teal', edgecolor='black', alpha=0.7, width=0.9, align='center')
plt.xlabel('Item Type')
plt.ylabel('Total Units Sold')
plt.title('Total Units Sold by Item Type')
plt.xticks(rotation=90)
# plt.yticks(rotation=45, fontsize=8)
plt.show()
```

```python
## plot total sales(unit sold) by item type 
import matplotlib.pyplot as plt

# Group the data by 'Item Type' and sum the 'Quantity Sold'
item_sales2 = sales_data_df.groupby('Item Type')['Units Sold'].sum()

item_sales2.plot(kind='bar', color='purple', edgecolor='black', alpha=0.7, width=0.7)
plt.xlabel('Item Type')
plt.ylabel('Total Units Sold')
plt.title('Total Units Sold by Item Type')
plt.xticks(rotation=90)
# plt.yticks(rotation=45, fontsize=8)
plt.show()
```
