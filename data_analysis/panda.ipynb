Handling missing values

```python
 #to check wheether there is missing value
df=pd.read_csv("employees.csv")
df.isnull().head(10)
```

```python
#checks column wise if there is null in any column .for the column has null=> True 
df.isnull().any()
```

```python
#checks row wise --> which row has null=> True 
df.isnull().any(axis=1)
```

```python
# to fill the null values 
df_filled=df.fillna("NA")
## fillna replaces all the NaN values with the specified value
```

```python
df
```

```python
df_filled
```

```python
#to populate the null with the mean of the column //// column-wise
df["SALARY_fillna"]=df["SALARY"].fillna(df["SALARY"].mean())
df
```

```python
df["SALARY"].mean()
```

```python
df.head(3)
```

```python
df.head(3).isnull()
```

```python
#Renaming columns
df=df.rename(columns={'EMPID':'EMP_ID',"EMAIL":"MAIL"})
df.head(3)
```

```python
# change the datatype of a column
df["SALARY"]=df["SALARY"].fillna(df["SALARY"].mean()).astype(int)
```

```python
df
```

```python
df["new_salary"]=df["SALARY"]*2
```

The other way to apply a particular function to every element of a column 

```python
df["new_salary"]=df["SALARY"].apply(lambda x:x*3)
```

```python
df.head(3)
```

```python
#grouping data bassed on values of a particular column
grouped_df=df.groupby("DEPARTMENT_ID")['SALARY'].mean()
print(grouped_df)
```

```python
c=df.groupby("DEPARTMENT_ID",).SALARY.sum()
print(c)
df.head(3)
```

```python
c=df.groupby(["DEPARTMENT_ID","MANAGER_ID"]).SALARY.sum()
print(c)
```

```python
# multiple aggregate function
agg=df.groupby("DEPARTMENT_ID").SALARY.agg(['mean','sum','count'])
print(agg)
```

Merging dataframes

```python
#it is like join operation of sql database
data1={
    "Key":['a','b','c'],
    "value1":[1,2,3]
}
data2={
    "Key":['a','b','d'],
    "value2":[2,3,4]
}
df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)
print(df1,"\n",df2)
```

```python
pd.merge(df1,df2,on="Key",how="inner")
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```
