```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
###Basic plotting
tips = sns.load_dataset('tips')
tips
```

```python
#Create a scatter plot
sns.scatterplot(data=tips, x='total_bill', y='tip',color='green', marker='o')
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python
##Line plot
sns.lineplot(data=tips, y='total_bill', x='size', color='red')
plt.title('Line Plot of Total Bill vs Size')
```

```python
#Catagorical plots
sns.barplot(data=tips, x='day', y='total_bill')
plt.title('Bar Plot of Total Bill by Day')
plt.show
```

```python
##box plot
sns.boxplot(data=tips, x='day', y='total_bill')
```

```python
#violin plot
sns.violinplot(data=tips, x='day', y='total_bill')
```

```python
#histogram
sns.histplot(data=tips, x='total_bill', bins=10,kde=True )
plt.title('Histogram of Total Bill')
```

```python
#KDE plot
sns.kdeplot(data=tips, x='total_bill', fill=True, color='purple')
##KDE plot
##sns.kdeplot(tips['total_bill'], fill=True, color='purple')
```

```python
##Pair plot
sns.pairplot(tips)
```

```python
#Heatmap
corr = tips[['total_bill', 'tip', 'size']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
corr
```

```python
import pandas as pd
data = pd.read_csv('100 Sales Records.csv')
data.head()
```

```python
plt.figure(figsize=(12,6))
sns.barplot(data=data, x='Item Type', y='Units Sold', estimator=sum)
plt.title('Total Units Sold by Item Type')
plt.xticks(rotation=90) # Rotate x-axis labels for better readability
plt.xlabel('Item Type')
plt.ylabel('Total Units Sold')

plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.barplot(data=data, x='Region', y='Units Sold', estimator=sum)
plt.title('Total Units Sold by Item Type')
plt.xticks(rotation=90) # Rotate x-axis labels for better readability
plt.xlabel('Item Type')
plt.ylabel('Total Units Sold')

plt.show()
```
