```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
%matplotlib inline
```

```python
from sklearn import datasets
iris=datasets.load_iris()
```

```python
iris_data = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_iris_data = scaler.fit_transform(iris_data)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
pca_iris = pca.fit_transform(scaled_iris_data)
```

```python
plt.scatter(pca_iris[:,0], pca_iris[:,1],c=iris.target, cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Iris Dataset - Hierarchical Clustering')
plt.show()
```

```python
##aglomerative clustering
import scipy.cluster.hierarchy as sc
##plot dendrogram
plt.figure(figsize=(20, 7))
plt.title("Dendrograms")
plt.xlabel("Samples")
plt.ylabel("Euclidean distances")
sc.dendrogram(sc.linkage(pca_iris, method="ward"))
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering,divisive_clustering
cluster = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
cluster.fit(pca_iris)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_iris[:,0], pca_iris[:,1],c=cluster.labels_, cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Iris Dataset - Hierarchical Clustering')
plt.show()
```
