```python
wordCount={}
charactersToRemove=","
with open("example.txt",'r') as exampleFile:
    for line in exampleFile:
        words=line.strip().split(" ")
        for word in words:
            word=word.lower()
            word=word.replace(charactersToRemove,"")
            if word in wordCount:
                wordCount[word]+=1
            else:
                wordCount[word]=1
            
        ##pass
   ## print(exampleFile)
for key,value in wordCount.items():
    print(f"{key}:{value}")

```

```python
with open ("example2.txt",'a') as writeFile:
    writeFile.write("Hello World!\n")
    writeFile.write("Hello Krishna!\n")
##with open ("example2.txt",'w') as writeFile:
    writeFile.write("Hello World!\n") 
with open("example2.txt",'r') as newFile:
     print(newFile.read())
```

```python
lines=("first line\n","second line\n","third line\n")
with open ("example2.txt",'a') as writeFile:
    writeFile.writelines(lines)
with open("example2.txt",'r') as newFile:
     print(newFile.read())
```

Read from a source file and write it in a destination file

```python
with open("example.txt",'r') as source:
    for line in source:
        with open("destination.txt",'a') as dest:
            dest.write(line)
```
