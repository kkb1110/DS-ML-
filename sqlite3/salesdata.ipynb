```python
import sqlite3 
connection=sqlite3.connect('sales_data.db')
cursor=connection.cursor()
```

```python
cursor.execute('''CREATE TABLE IF NOT EXISTS sales
               (id INTEGER PRIMARY KEY,
               product_name TEXT not null,
               quantity INTEGER not null,
               price REAL not null,
               date TEXT not null)
               ''')
#price REAL not null, here REAL is used to store decimal values

```

```python
sales_data = [
    (1, 'Laptop', 5, 800.00, '2024-01-15'),
    (2, 'Smartphone', 10, 500.00, '2024-01-16'),
    (3, 'Tablet', 7, 300.00, '2024-01-17'),
    (4, 'Headphones', 15, 50.00, '2024-01-18'),
    (5, 'Smartwatch', 8, 200.00, '2024-01-19')
]

cursor.executemany('''
INSERT INTO sales (id, product_name, quantity, price, date)
VALUES (?, ?, ?, ?, ?)
''', sales_data)

connection.commit()
```

```python
cursor.execute('''SELECT * FROM sales''')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
connection.close()
```
