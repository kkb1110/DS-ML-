```python
connection = sql.connect('test.db')
cursor = connection.cursor()
```

```python
cursor.execute('''
               CREATE TABLE IF NOT EXISTS employees (
               id INTEGER PRIMARY KEY,
               name TEXT NOT NULL,
               age INTEGER NOT NULL,
               department TEXT NOT NULL)
              ''' )
```
