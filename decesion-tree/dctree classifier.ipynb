```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
print(iris['data'])
```

```python
X=pd.DataFrame(iris['data'],columns=iris['feature_names'])
y=np.array(iris['target'])
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=91)
```

```python
#apply decision tree classifier
from sklearn.tree import DecisionTreeClassifier
dctreeClassifier=DecisionTreeClassifier(max_depth=2)
dctreeClassifier.fit(X_train,y_train)
```

```python
## visualize the tree
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(dctreeClassifier,filled=True)
```

```python
y_pred=dctreeClassifier.predict(X_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

## Prepruning & Hyperparameter tuninng

```python
params={
    'criterion':['gini','entropy','log_loss'],
    'splitter':['best','random'],
    'max_depth':[2,3,4,5],
    'max_features':['auto','sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator=dctreeClassifier,param_grid=params,cv=5,scoring='accuracy')
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
from sklearn.metrics import accuracy_score
y_pred=grid.predict(X_test)
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred)*100,'%')
```
